```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.layers import Dense, Conv2D, Input, Flatten
from keras import Sequential
from keras.datasets import mnist
```

```python
(x_train,y_train),(x_test,y_test)= mnist.load_data()
```

```python
model = Sequential()
model.add(Input(shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding="valid",activation="relu"))
model.add(Conv2D(32,kernel_size=(3,3),padding="valid",activation="relu"))
model.add(Conv2D(32,kernel_size=(3,3),padding="valid",activation="relu"))
model.add(Flatten())
model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation="softmax"))
```

The way we have a dense layer in the ANN, for the CNN we have Conv2D layers. For the first layer and the other layer we are using 32 filters of size 3x3, without padding for now (valid), and the activation layer of ReLU. That's it. Then we are flattening that array and sending it to the neural network and getting the output.

```python
model.summary()
```

We can observe the reducing size of my image:
1. 28 × 28
2. 26 × 26
3. 24 × 24
4. 22 × 22, and so on
The reducing size of the image is of the feature map without padding so information loss can happen to the corner pixels. To overcome it we are using padding.

```python
model = Sequential()
model.add(Input(shape=(28,28,1)))
model.add(Conv2D(32,kernel_size=(3,3),padding="same",activation="relu"))
model.add(Conv2D(32,kernel_size=(3,3),padding="same",activation="relu"))
model.add(Conv2D(32,kernel_size=(3,3),padding="same",activation="relu"))
model.add(Flatten())
model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation="softmax"))
```

Here we are using padding

padding = "same"

```python
model.summary()
```

We can observe that the feature map still remains 28x28 so there is no loss of my image information while having convolutional operations.
