```python
from tensorflow.keras.applications.resnet50 import ResNet50
from tensorflow.keras.preprocessing import image
from tensorflow.keras.applications.resnet50 import preprocess_input,decode_predictions
import numpy as np
```

```python
model = ResNet50(weights="imagenet")
```

```python
image_path="/content/images (4).jpg"
img = image.load_img(image_path,target_size=(224,224))
x = image.img_to_array(img)
x =np.expand_dims(x,axis=0)
x=preprocess_input(x)
```

```python
preds= model.predict(x)
print("Predicted",decode_predictions(preds,top=5)[0])
```

Given the prediction of the top 5, what can it inherit or think about what the data is on the pre-trained model ResNet50? ResNet50 is a pre-trained model on the ImageNet dataset and won in 2016 for having a 3.5% loss only. Which means a human can make an error of 5%, whereas this model is making an error of only 3.5%. This is to be considered the most widely used and famous pre-trained model.
