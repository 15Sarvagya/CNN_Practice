```python
import tensorflow
from tensorflow import keras
from tensorflow.keras.layers import Dense,Input,Conv2D,Flatten,AveragePooling2D
from tensorflow.keras import Sequential
from keras.datasets import mnist
```

```python
(x_train,y_train),(x_test,y_test)= mnist.load_data()
```

```python
model = Sequential()
model.add(Input(shape=(32,32,1)))

model.add(Conv2D(6,kernel_size=(5,5),padding="valid",activation="tanh"))
model.add(AveragePooling2D(pool_size=(2,2),strides=2,padding="valid"))

model.add(Conv2D(16,kernel_size=(5,5),padding="valid",activation="tanh"))
model.add(AveragePooling2D(pool_size=(2,2),strides=2,padding="valid"))

model.add(Flatten())

model.add(Dense(120,activation="tanh"))

model.add(Dense(84,activation="tanh"))

model.add(Dense(10,activation="softmax"))
```

Here is the architecture of LeNet-5 because it covers the five layers:
1. The convolutional operation on 6 filters of size 5 × 5, applying average pooling.
2. The convolutional operation On 16 filters of 5 × 5 size, with average pooling again.
3. Adding 120 neurons
4. Adding 84 neurons
5. Output layer of 10 outputs


The thing that we can observe from this is that the more depth we have in the convolutional operation, the more filters we are using. At the time the LeNet-5 architecture was discovered, the best optimizer was 10H. That's why we are using 10H.
